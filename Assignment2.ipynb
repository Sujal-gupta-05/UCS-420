```python
import random
random.seed(int(rollno))


numbers = []

for i in range(100):
    numbers.append(random.randint(100, 900))

print(numbers)
```

```python
odd_nums = [num for num in numbers if num % 2 != 0]
print(odd_nums)
```

```python
even_nums = [num for num in numbers if num % 2 == 0]
print(even_nums)
```

```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

print(is_prime(10))
```

```python
prime_nums = [num for num in numbers if isPrime(num)]
print(prime_nums)
```

```python
count = len(prime_nums)
print(count)
```

```python
from collections import Counter
frequency = Counter(numbers)
most_common_number, occurrence = frequency.most_common(1)[0]

print(most_common_number)
print(occurrence)
```

# Q4

```python
rollno = "1024170207"

digits = [int(c) for c in rollno]

A = {digit * 7 for digit in digits}
B = {digit * 9 for digit in digits}

print(A)
print(B)
```

```python
union = A.union(B)
intersection = A.intersection(B)
print(union)
print(intersection)
```

```python
A_minus_B = A.difference(B)
B_minus_A = B.difference(A)

print(A_minus_B)
print(B_minus_A)
```

```python
# difference() gives elements present in the first set but not the second 
# symmetric_difference() gives elements present in either set but not both.

print(A.symmetric_difference(B))
```

```python
print(A.issubset(B))
print(B.issuperset(A))
```

```python
X = int(input("Enter a value X: "))
A.discard(X)
print(A)

# discard() does not raise an error if the element is not present, unlike remove().
```

```python
my_dict = {
    "name": "Sujal",
    "roll_no": "1024170207",
    "branch": "CSE",
    "age": 21,
    "city": "Jammu"
}

my_dict["location"] = my_dict.pop("city")
print(my_dict)
```

```python
my_dict["cgpa"] = 9.21
my_dict["age"] = my_dict["age"] + 1

dict_pop = my_dict.copy()
dict_del = my_dict.copy()

dict_pop.pop("branch")
del dict_del["branch"]

print("Using pop():", dict_pop)
print("Using del:", dict_del)

# pop() removes the key and returns its value, whereas del removes the key but returns nothing.
```

```python
for key, value in my_dict.items():
    print(key, "→", value)
```

```python
if "email" in my_dict:
    print("Email:", my_dict["email"])
else:
    print("Email key does not exist.")
```

```python
friend_dict = {
    "name": "Rahul",
    "roll_no": "87654321",
    "branch": "CSE",
    "age": 21,
    "city": "Mumbai"
}

merged_dict = {**my_dict, **friend_dict}

print("Merged dictionary:", merged_dict)

# When both dictionaries have the same key, the value from the second dictionary wins.
```
