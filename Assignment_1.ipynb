```python
# 5.3
n = int(input("Enter the number: "))
sum = 0
for i in range(2, n + 1):
    is_prime = True
    for j in range(2, int(i**0.5) + 1):
        if i % j == 0:
            is_prime = False
            break
    if is_prime:
        sum += i

print(sum)
```

```python
#6.1
def AddOdd(n):
    sum = 0
    
    for i in range(1, n + 1):
        if i % 2 != 0:
            sum += i
            
    return sum


n = int(input("Enter n: "))
print("Sum of odd numbers =", AddOdd(n))
```

```python
#6.2
def IsPrime(n):
    if n < 2:
        return False
    
    for i in range(2, n):
        if n % i == 0:
            return False
    
    return True


def AddPrime(n):
    sum = 0
    
    for i in range(1, n + 1):
        if IsPrime(i):
            sum += i
            
    return sum


n = int(input("Enter n: "))
print("Sum of prime numbers =", AddPrime(n))
```

### **8.1 Indexing in string**

```python
var = 'Sujal gupta!'
print ("var      = ", var)
print ("var[0]   = ", var[0])
print ("var[1:5] = ", var[1:5])
print ("var[:-5] = ", var[:-5])
```

---
### **8.2 String length, upper, lower**
---

```python
var = 'Sujal Gupta'
print ("String = ", var)
print ("Length = : ", len(var))
print ("Upper  = : ", var.upper())
print ("Lower  = : ", var.lower())
```

### **8.5 String strip**
---

```python
var =" Sujal   Gupta    "

print("String    = ", var)
print("Length    = ", len(var))
print("var strip = ", var.strip())
print("Length of var after strip = ", len(var.strip()))
```

---
### **8.6 String split**
---

```python
var =" Sujal,   Gupta    "

print("String    = ", var)
print("Length    = ", len(var))
print("var split = ", var.split())
print("var split = ", var.split(' '))
print("var split = ", var.split(','))

print("var split = ", var.strip().split(','))
```

---
### **8.7 Count in string**
---

```python
var=" Indian Army    "
print ("String       = ", var)
print ("Count of ' ' = ", var.count(' '))
print ("Count of 'a' = ", var.count('a'))
print ("Count of 'n' = ", var.count('an'))
```

---
### **8.8 Reverse a String**
---

```python
var="Palindrome"
print ("String    = ", var)
print ("var[::1]  = ", var[::1])
print ("var[::2]  = ", var[::2])
print ("var[::-1] = ", var[::-1])
print ("var[::-2] = ", var[::-2])

var=var[::-1]
print ("var after reverse = ", var)
```
