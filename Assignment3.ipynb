```python
emp_data = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Edward"],
    "Department": ["HR", "IT", "IT", "Marketing", "Sales"],
    "Age": [29, 34, 41, 28, 38],
    "Salary": [50000, 70000, 65000, 55000, 60000],
    "Years_of_Experience": [4, 8, 10, 3, 12],
    "Joining_Date": ["2020-03-15", "2017-07-19", "2013-06-01", "2021-02-10", "2010-11-25"],
    "Gender": ["Female", "Male", "Male", "Female", "Male"],
    "Bonus": [5000, 7000, 6000, 4500, 5000],
    "Rating": [4.5, 4.0, 3.8, 4.7, 3.5]
}

emp_df = pd.DataFrame(emp_data)
emp_df.to_csv("employees.csv", index=False)

emp = pd.read_csv("employees.csv")
emp
```

```python
print("Shape of the dataset (rows, columns):", emp.shape)
```

```python
emp.info()
```

```python
emp.describe()
```

```python
print("First 5 rows:")
print(emp.head())

print("\nLast 3 rows:")
print(emp.tail(3))
```

```python
avg_salary = emp["Salary"].mean()
print("Average Salary:", avg_salary)

total_bonus = emp["Bonus"].sum()
print("Total Bonus paid:", total_bonus)

youngest_age = emp["Age"].min()
print("Youngest employee age:", youngest_age)


highest_rating = emp["Rating"].max()
print("Highest Rating:", highest_rating)
```

```python
sorted_emp = emp.sort_values(by="Salary", ascending=False)
sorted_emp
```

```python
def get_performance(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4.0:
        return "Good"
    else:
        return "Average"

emp["Performance"] = emp["Rating"].apply(get_performance)
emp
```

```python
emp.isnull().sum()
```

```python
emp = emp.rename(columns={"Employee_ID": "ID"})
emp.head()
```

```python
exp_more_than_5 = emp[emp["Years_of_Experience"] > 5]
print("Employees with more than 5 years experience:")
print(exp_more_than_5)


it_dept = emp[emp["Department"] == "IT"]
print("\nEmployees in IT department:")
print(it_dept)
```

```python
emp["Tax"] = emp["Salary"] * 0.10
emp
```

```python
emp.to_csv("employees_modified.csv", index=False)
print("Saved successfully as employees_modified.csv")
```
